Q4. EXTRACTING TEXT FROM PAGE NUMBER 2 OF "BUSINESS_PROPOSAL.PDF" USING "PyPDF2".

```python
pip install PyPDF2 
```

```python
import PyPDF2

file = open(R"C:\Users\HELLO\Downloads\Business_Proposal.pdf" , 'rb')
pdf_reader = PyPDF2.PdfReader(file)

page_2_text = pdf_reader.pages[1].extract_text()

file.close()
print(page_2_text)
```

Q5. APPENDING PAGE 2 TEXT TO "CONTACTS.TXT".

```python
with open('contacts.txt' , 'a+') as f:
    f.write(page_2_text)

with open('contacts.txt' , 'r') as f:
    print(f.read())
```

Q6. EXTRACTING EMAIL ADDRESSES FROM "PAGE_2_TEXT".

```python
import re

pattern = r'[\w\.-]+@[\w\.-]+'

emails = re.findall(pattern , page_2_text)
print(emails)
```
